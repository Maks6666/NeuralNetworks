LSTM - long short -term memory - type of recurrent neural network for memorizing and analyzing time sequences. Designed to prevent such common issues of recurrent networks as gradient vanishing or gradient explosion.  

```python
import torch 
from torch import nn 
```

```python
conv_net = nn.Sequential(
    nn.Conv2d(in_channels=3, out_channels=32, kernel_size=3, stride=2),
    nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3, stride=2),
    nn.Conv2d(in_channels=64, out_channels=128, kernel_size=3, stride=2),
    nn.Conv2d(in_channels=128, out_channels=256, kernel_size=3, stride=2),
    nn.Conv2d(in_channels=256, out_channels=512, kernel_size=3, stride=2),
    nn.Conv2d(in_channels=512, out_channels=1024, kernel_size=3, stride=2),
    
)
```

```python
# to extract visual features, before passing through LSTM block, tensor should be passed through
# convolutional blocks.

# imagine like we've got a video clip with 16, of size 128 x 128 and in RGB-format, let it be a tensor:
tensor = torch.randn(1, 16, 3, 128, 128)
B, F, C, H, W = tensor.shape
# B - batch
# F - frame
# C - channel
# H, W - height, width
out = tensor.view(B*F, C, H, W)
out = conv_net(out)
out = out.flatten(1)
out = out.view(B, F, -1)
out.shape # -> torch.Size([1, 16, 1024]) - this already could be passed through LSTM
```

```python
B, F, C = out.shape # -> [1, 16, 1024]

lstm = nn.LSTM(
    input_size=C,
    hidden_size=512,
    num_layers=1,
    batch_first=True,
    bidirectional=True,
)

output, (h, c) = lstm(out)
print(out.shape, h.shape, c.shape)

# LSTM block in PyTorch is not only a block, but a sequence of LSTM cells. So if we've got 16 frames in video, we will have 16 cells of LSTM in a network

# -> (torch.Size([1, 16, 1024]), torch.Size([2, 1, 512]), torch.Size([2, 1, 512]))

# where output - is a collection of hidden states (short memory) of each frame - thats why it has 16 frames. 
# It also has a size of 1024 (512 * 2) because block is bidirectional and contains forward and back states 

# h - is a short memory on step t
# c - is a long memory on step t 

# both of them have size of [num_layers * num_directions, B, hidden_size]
```
